# Combine Performance and Monitoring

This includes filtering for the first client and incorporating efficiency metrics to track performance relative to resource usage.

```python
client = 1

df_performance_monitoring = collect.get_monitoring_aggregated_per_phase_multitenant(type="stream")
df_performance = collect.get_performance_aggregated_per_phase_multitenant()
join_keys = ["code", "experiment_run", "client", "type_tenants", "num_tenants"]
cols_to_use = [c for c in df_performance_monitoring.columns if c not in df_performance.columns]
df_perf_k = df_performance.copy()
df_mon_k = df_performance_monitoring[join_keys + cols_to_use].copy()
for k in join_keys:
    if df_perf_k[k].dtype != df_mon_k[k].dtype:
        df_perf_k[k] = df_perf_k[k].astype(str)
        df_mon_k[k] = df_mon_k[k].astype(str)
merged_df = df_perf_k.merge(df_mon_k, on=join_keys, how="inner")
merged_df = collect.add_metadata(merged_df).copy()
for col in merged_df.columns:
    if merged_df[col].dtype == object:
        merged_df[col] = pd.to_numeric(merged_df[col], errors='coerce').fillna(merged_df[col])

#merged_df = merged_df[merged_df["client"] == client]

merged_df["RAM/Throughput"] = merged_df["Memory Usage Cached [MiB]"] / (merged_df["Throughput@Size"])
merged_df["CPUs/Throughput"] = merged_df["CPU Utilization"] / (merged_df["Throughput@Size"] / 3600.)
#merged_df["E_RAM"] = (merged_df["Throughput@Size"]) / merged_df["Max RAM [Gb]"]
#merged_df["E_CPU"] = (merged_df["Throughput@Size"] * 3600.) / merged_df["CPU [CPUs]"]
merged_df["E_CPU"] = (10. * merged_df["num_tenants"] * merged_df["num_of_queries"] * merged_df["SF"]) / merged_df["CPU Utilization"]
merged_df["E_RAM"] = (10. * merged_df["num_tenants"] * merged_df["num_of_queries"] * merged_df["SF"]) / merged_df["Memory Usage Cached [MiB]"]

merged_df.T
```

## With Metadata

Connection infos are added.

```python
df = collect.add_metadata(merged_df)
df.T
```
